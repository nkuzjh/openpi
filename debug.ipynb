# 可视化openpi原有lr_schedule

这个 lr_scheduler 实现的是一个典型的 “线性预热 (Linear Warmup) + 余弦退火 (Cosine Decay)” 策略。

它的变化过程分为两个阶段：
    线性预热阶段 (Warmup Phase)：
        范围：从第 0 步 到 warmup_steps。
        行为：学习率从一个较小的初始值 init_lr 线性增加 到 peak_lr（峰值学习率）。
        特点：代码中有一个细节注释 Match JAX behavior，说明它不是从 0 开始，而是从 peak_lr / (warmup_steps + 1) 这个值开始爬升。
    余弦衰减阶段 (Cosine Decay Phase)：
        范围：从 warmup_steps 到 decay_steps。
        行为：学习率从 peak_lr 开始，按照余弦曲线（Cosine Curve） 平滑下降，最终在 decay_steps 结束时降至 end_lr（最小学习率）。
        特点：下降速度是“先慢、中间快、后慢”，形成一个平滑的 S 形下坡。

运行结果预期
    红线左侧：你会看到一条直直的斜线向上爬升（这就是线性预热）。
    红线右侧：你会看到一条平滑的曲线向下弯曲，直到接近 0 或设定的 end_lr（这就是余弦衰减）。

```python
import numpy as np
import matplotlib.pyplot as plt

# ================= 配置参数 (模拟你的环境) =================
peak_lr = 0.001       # 最高学习率
end_lr = 0.00001      # 最低学习率 (衰减终点)
warmup_steps = 100    # 预热步数
decay_steps = 1000    # 总衰减步数 (通常对应总训练步数)

# ================= 你的 lr_scheduler 函数 =================
def lr_schedule(step: int):
    # 阶段 1: 线性预热
    if step < warmup_steps:
        # Match JAX behavior: start from peak_lr / (warmup_steps + 1)
        init_lr = peak_lr / (warmup_steps + 1)
        return init_lr + (peak_lr - init_lr) * step / warmup_steps

    # 阶段 2: 余弦衰减
    progress = min(1.0, (step - warmup_steps) / max(1, decay_steps - warmup_steps))
    cos = 0.5 * (1 + np.cos(np.pi * progress))
    return end_lr + (peak_lr - end_lr) * cos

# ================= 可视化逻辑 =================
def plot_scheduler():
    # 生成模拟的步数 (从 0 到 decay_steps，多画一点看看结尾)
    steps = np.arange(0, decay_steps + 50)

    # 计算每一步的 LR
    lrs = [lr_schedule(s) for s in steps]

    # 绘图
    plt.figure(figsize=(10, 6))
    plt.plot(steps, lrs, label='Learning Rate', color='blue', linewidth=2)

    # 标记关键点
    plt.axvline(x=warmup_steps, color='red', linestyle='--', alpha=0.6, label='Warmup End')
    plt.axhline(y=peak_lr, color='green', linestyle=':', alpha=0.6, label='Peak LR')
    plt.axhline(y=end_lr, color='orange', linestyle=':', alpha=0.6, label='End LR')

    # 图表装饰
    plt.title('Linear Warmup + Cosine Decay Schedule', fontsize=14)
    plt.xlabel('Iterations (Steps)', fontsize=12)
    plt.ylabel('Learning Rate', fontsize=12)
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.show()

if __name__ == "__main__":
    plot_scheduler()
```

## 不同模块学习率

```python
import torch
import torch.nn as nn

def get_optimizer_params(model: nn.Module, config: dict):
    """
    为优化器构建参数组，区分新模块/骨干网络，以及是否应用权重衰减。
    """
    # 1. 获取超参数
    new_module_lr = config.get('new_module_lr', 1e-4)
    backbone_lr = config.get('backbone_lr', 2e-5)
    weight_decay = config.get('weight_decay', 0.05)

    # 目标新模块的关键词标识
    new_module_keyword = "fps_guided_map_attention"

    # 2. 初始化四个参数桶
    # 组1: 新模块 & 需要衰减 (Linear weights, Conv weights)
    params_new_decay = []
    # 组2: 新模块 & 不需要衰减 (Bias, LayerNorm)
    params_new_no_decay = []
    # 组3: 骨干网络 & 需要衰减
    params_backbone_decay = []
    # 组4: 骨干网络 & 不需要衰减
    params_backbone_no_decay = []

    # 不需要权重衰减的关键词黑名单
    no_decay_keywords = {'bias', 'norm', 'embedding', 'absolute_pos_embed', 'relative_position_bias_table'}

    # 3. 遍历参数并分类
    for name, param in model.named_parameters():
        if not param.requires_grad:
            continue

        # --- 判断 A: 是否应用 Weight Decay ---
        # 规则: 维度<2 (通常是bias或1D tensor) 或者名字里包含黑名单关键词的，不进行衰减
        if param.ndim < 2 or any(wd_key in name for wd_key in no_decay_keywords):
            apply_decay = False
        else:
            apply_decay = True

        # --- 判断 B: 属于新模块还是骨干网络 ---
        is_new_module = new_module_keyword in name

        # --- 分发到对应的桶 ---
        if is_new_module:
            if apply_decay:
                params_new_decay.append(param)
            else:
                params_new_no_decay.append(param)
        else:
            if apply_decay:
                params_backbone_decay.append(param)
            else:
                params_backbone_no_decay.append(param)

    # 4. 构建优化器输入列表
    optim_groups = [
        {
            'params': params_new_decay,
            'lr': new_module_lr,
            'weight_decay': weight_decay,
            'name': 'new_module_decay'
        },
        {
            'params': params_new_no_decay,
            'lr': new_module_lr,
            'weight_decay': 0.0,
            'name': 'new_module_no_decay'
        },
        {
            'params': params_backbone_decay,
            'lr': backbone_lr,
            'weight_decay': weight_decay,
            'name': 'backbone_decay'
        },
        {
            'params': params_backbone_no_decay,
            'lr': backbone_lr,
            'weight_decay': 0.0,
            'name': 'backbone_no_decay'
        },
    ]

    # 5. (可选) 打印统计信息，防止因为名字匹配错误导致参数漏掉或分错
    print(f"Optimizer groups summary:")
    print(f"  - New Module (LR={new_module_lr}): {len(params_new_decay)} decay params, {len(params_new_no_decay)} no-decay params")
    print(f"  - Backbone   (LR={backbone_lr}): {len(params_backbone_decay)} decay params, {len(params_backbone_no_decay)} no-decay params")

    return optim_groups



# 初始化优化器
optimizer_grouped_parameters = get_optimizer_params(model, config)
optimizer = torch.optim.AdamW(optimizer_grouped_parameters) # 注意这里不需要再传 lr 或 weight_decay，因为都在 groups 里定义了
```
